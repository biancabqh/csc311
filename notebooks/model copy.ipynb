# Random Forest

```python
# Model 
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)

rf_features = ['difficulty_clean',
       'strategy_importance_num', 'injury_risk_comfort_num',
       'individual_brilliance_num', 'play_season_Fall', 'play_season_None',
       'play_season_Spring', 'play_season_Summer', 'play_season_Winter',
       'play_location_At home or nearby',
       'play_location_Community recreation facility',
       'play_location_Indoor facility', 'play_location_None',
       'play_location_Others', 'play_location_Outdoor public space',
       'play_location_Private sports facility',
       'play_location_School or university facility',
       'associated_music_Classical or jazz', 'associated_music_Country',
       'associated_music_Electronic or dance',
       'associated_music_Folk or traditional',
       'associated_music_Hip-hop or rap', 'associated_music_Latin',
       'associated_music_No particular type', 'associated_music_None',
       'associated_music_Other', 'associated_music_Pop',
       'associated_music_Rock', 'associated_foods_drinks_is_missing',
       'associated_foods_drinks_tfidf_beer',
       'associated_foods_drinks_tfidf_dogs',
       'associated_foods_drinks_tfidf_hot',
       'associated_foods_drinks_tfidf_popcorn',
       'associated_foods_drinks_tfidf_soda', 'associated_color_is_missing',
       'associated_color_tfidf_blue',
       'associated_color_tfidf_green', 'associated_color_tfidf_orange',
       'associated_color_tfidf_red', 'associated_color_tfidf_white',
       'associated_song_is_missing', 'associated_song_tfidf_ball', 'associated_song_tfidf_game',
       'associated_song_tfidf_missing', 'associated_song_tfidf_shakira',
       'associated_song_tfidf_waka', 'emotional_response_is_missing',
       'emotional_response_tfidf_bored',
       'emotional_response_tfidf_dont', 'emotional_response_tfidf_excited',
       'emotional_response_tfidf_excitement', 'emotional_response_tfidf_feel', 'log_price']

rt_target = 'target_sport_num'

X_train_rt = train_df[rf_features]
Y_train_rt = train_df[rt_target]

rf.fit(X_train_rt, Y_train_rt)
```

```python
# Training

print("================== Random Forest Training Metrics ==================")

y_pred = rf.predict(train_df[rf_features])
y_true = train_df[rt_target]

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="macro")
recall = recall_score(y_true, y_pred, average="macro")
f1 = f1_score(y_true, y_pred, average="macro")

print(f"Training Accuracy:        {accuracy:.4f}")
print(f"Training Macro Precision: {precision:.4f}")
print(f"Training Macro Recall:    {recall:.4f}")
print(f"Training Macro F1:        {f1:.4f}")  
print("Confusion Matrix:")
print(confusion_matrix(train_df[rt_target], y_pred))
print("Classification Report:")
print(classification_report(train_df[rt_target], y_pred))
```

```python
# Validation

print("================== Random Forest Validation Metrics ==================")

y_pred = rf.predict(val_df[rf_features])
y_true = val_df[rt_target]

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="macro")
recall = recall_score(y_true, y_pred, average="macro")
f1 = f1_score(y_true, y_pred, average="macro")

print(f"Validation Accuracy:        {accuracy:.4f}")
print(f"Validation Macro Precision: {precision:.4f}")
print(f"Validation Macro Recall:    {recall:.4f}")
print(f"Validation Macro F1:        {f1:.4f}")  
print("Confusion Matrix:")
print(confusion_matrix(val_df[rt_target], y_pred))
print("Classification Report:")
print(classification_report(val_df[rt_target], y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer

lr = LogisticRegression(max_iter=1000, random_state=42)

lr_features = ['price_clean', 'difficulty_clean',
       'strategy_importance_num', 'injury_risk_comfort_num',
       'individual_brilliance_num', 'play_season_Fall', 'play_season_None',
       'play_season_Spring', 'play_season_Summer', 'play_season_Winter',
       'play_location_At home or nearby',
       'play_location_Community recreation facility',
       'play_location_Indoor facility', 'play_location_None',
       'play_location_Others', 'play_location_Outdoor public space',
       'play_location_Private sports facility',
       'play_location_School or university facility',
       'associated_music_Classical or jazz', 'associated_music_Country',
       'associated_music_Electronic or dance',
       'associated_music_Folk or traditional',
       'associated_music_Hip-hop or rap', 'associated_music_Latin',
       'associated_music_No particular type', 'associated_music_None',
       'associated_music_Other', 'associated_music_Pop',
       'associated_music_Rock', 'associated_foods_drinks_is_missing',
       'associated_foods_drinks_tfidf_beer',
       'associated_foods_drinks_tfidf_dogs',
       'associated_foods_drinks_tfidf_hot',
       'associated_foods_drinks_tfidf_popcorn',
       'associated_foods_drinks_tfidf_soda', 'associated_color_is_missing',
       'associated_color_tfidf_blue',
       'associated_color_tfidf_green', 'associated_color_tfidf_orange',
       'associated_color_tfidf_red', 'associated_color_tfidf_white',
       'associated_song_is_missing', 'associated_song_tfidf_ball', 'associated_song_tfidf_game',
       'associated_song_tfidf_missing', 'associated_song_tfidf_shakira',
       'associated_song_tfidf_waka', 'emotional_response_is_missing',
       'emotional_response_tfidf_bored',
       'emotional_response_tfidf_dont', 'emotional_response_tfidf_excited',
       'emotional_response_tfidf_excitement', 'emotional_response_tfidf_feel', 'log_price']

lr_target = 'target_sport_num'

X_train_lr = train_df[lr_features]
Y_train_lr = train_df[lr_target]

imputer = SimpleImputer(strategy='median')
X_train_lr_imputed = imputer.fit_transform(X_train_lr)
X_val_lr_imputed = imputer.transform(val_df[lr_features])
X_test_lr_imputed = imputer.transform(test_df[lr_features])

lr.fit(X_train_lr_imputed, Y_train_lr)

```

```python
# Training

print("================== Logistic Regression Training Metrics ==================")

y_pred = lr.predict(X_train_lr_imputed)
y_true = train_df[lr_target]

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="macro")
recall = recall_score(y_true, y_pred, average="macro")
f1 = f1_score(y_true, y_pred, average="macro")

print(f"Training Accuracy:        {accuracy:.4f}")
print(f"Training Macro Precision: {precision:.4f}")
print(f"Training Macro Recall:    {recall:.4f}")
print(f"Training Macro F1:        {f1:.4f}")  
print("Confusion Matrix:")
print(confusion_matrix(train_df[lr_target], y_pred))
print("Classification Report:")
print(classification_report(train_df[lr_target], y_pred))
```

```python
# Validation 

print("================== Logistic Regression Validation Metrics ==================")

y_pred = lr.predict(X_val_lr_imputed)
y_true = val_df[lr_target]

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="macro")
recall = recall_score(y_true, y_pred, average="macro")
f1 = f1_score(y_true, y_pred, average="macro")

print(f"Validation Accuracy:        {accuracy:.4f}")
print(f"Validation Macro Precision: {precision:.4f}")
print(f"Validation Macro Recall:    {recall:.4f}")
print(f"Validation Macro F1:        {f1:.4f}")  
print("Confusion Matrix:")
print(confusion_matrix(val_df[lr_target], y_pred))
print("Classification Report:")
print(classification_report(val_df[lr_target], y_pred))
```

# Naive Bayes

```python
# Multinomial Naive Bayes for the TF-IDF text features
from sklearn.naive_bayes import MultinomialNB

nb_features = [col for col in train_df.columns if "_tfidf_" in col]
nb_target = "target_sport_num"

nb = MultinomialNB(alpha=1.0)
nb.fit(train_df[nb_features], train_df[nb_target])
```

```python
# Training metrics
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
)

nb_train_pred = nb.predict(train_df[nb_features])
nb_train_true = train_df[nb_target]

print("================== Naive Bayes Training Metrics ==================")
print(f"Training Accuracy:        {accuracy_score(nb_train_true, nb_train_pred):.4f}")
print(f"Training Macro Precision: {precision_score(nb_train_true, nb_train_pred, average='macro'):.4f}")
print(f"Training Macro Recall:    {recall_score(nb_train_true, nb_train_pred, average='macro'):.4f}")
print(f"Training Macro F1:        {f1_score(nb_train_true, nb_train_pred, average='macro'):.4f}")
print("Confusion Matrix:")
print(confusion_matrix(nb_train_true, nb_train_pred))
print("Classification Report:")
print(classification_report(nb_train_true, nb_train_pred))
```

```python
# Validation metrics
nb_val_pred = nb.predict(val_df[nb_features])
nb_val_true = val_df[nb_target]

print("================== Naive Bayes Validation Metrics ==================")
print(f"Validation Accuracy:        {accuracy_score(nb_val_true, nb_val_pred):.4f}")
print(f"Validation Macro Precision: {precision_score(nb_val_true, nb_val_pred, average='macro'):.4f}")
print(f"Validation Macro Recall:    {recall_score(nb_val_true, nb_val_pred, average='macro'):.4f}")
print(f"Validation Macro F1:        {f1_score(nb_val_true, nb_val_pred, average='macro'):.4f}")
print("Confusion Matrix:")
print(confusion_matrix(nb_val_true, nb_val_pred))
print("Classification Report:")
print(classification_report(nb_val_true, nb_val_pred))
```

# Naive Bayes + Random Forest Ensemble

```python
# Soft-voting ensemble: average the class probabilities from both models
ensemble_rf_weight = 0.5
ensemble_nb_weight = 0.5

ensemble_train_probabilities = (
    ensemble_rf_weight * rf.predict_proba(train_df[rf_features])
    + ensemble_nb_weight * nb.predict_proba(train_df[nb_features])
)
ensemble_val_probabilities = (
    ensemble_rf_weight * rf.predict_proba(val_df[rf_features])
    + ensemble_nb_weight * nb.predict_proba(val_df[nb_features])
)

ensemble_train_pred = rf.classes_[ensemble_train_probabilities.argmax(axis=1)]
ensemble_val_pred = rf.classes_[ensemble_val_probabilities.argmax(axis=1)]
```

```python
# Ensemble training metrics
ensemble_train_true = train_df[rt_target]

print("================== Ensemble Training Metrics ==================")
print(f"Training Accuracy:        {accuracy_score(ensemble_train_true, ensemble_train_pred):.4f}")
print(f"Training Macro Precision: {precision_score(ensemble_train_true, ensemble_train_pred, average='macro'):.4f}")
print(f"Training Macro Recall:    {recall_score(ensemble_train_true, ensemble_train_pred, average='macro'):.4f}")
print(f"Training Macro F1:        {f1_score(ensemble_train_true, ensemble_train_pred, average='macro'):.4f}")
print("Confusion Matrix:")
print(confusion_matrix(ensemble_train_true, ensemble_train_pred))
print("Classification Report:")
print(classification_report(ensemble_train_true, ensemble_train_pred))
```

```python
# Ensemble validation metrics
ensemble_val_true = val_df[rt_target]

print("================== Ensemble Validation Metrics ==================")
print(f"Validation Accuracy:        {accuracy_score(ensemble_val_true, ensemble_val_pred):.4f}")
print(f"Validation Macro Precision: {precision_score(ensemble_val_true, ensemble_val_pred, average='macro'):.4f}")
print(f"Validation Macro Recall:    {recall_score(ensemble_val_true, ensemble_val_pred, average='macro'):.4f}")
print(f"Validation Macro F1:        {f1_score(ensemble_val_true, ensemble_val_pred, average='macro'):.4f}")
print("Confusion Matrix:")
print(confusion_matrix(ensemble_val_true, ensemble_val_pred))
print("Classification Report:")
print(classification_report(ensemble_val_true, ensemble_val_pred))
```
